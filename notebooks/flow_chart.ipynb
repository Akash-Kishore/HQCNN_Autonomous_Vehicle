```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

def draw_flowchart():
    # Setup the canvas
    fig, ax = plt.subplots(figsize=(12, 16))
    ax.set_xlim(0, 10)
    ax.set_ylim(0, 18)
    ax.axis('off')

    # Helper function to draw the process boxes
    def draw_box(x, y, text, color='white', width=4.5, height=1.4):
        rect = patches.FancyBboxPatch((x - width/2, y - height/2), width, height, 
                                     boxstyle="round,pad=0.2", linewidth=2, 
                                     edgecolor='black', facecolor=color)
        ax.add_patch(rect)
        ax.text(x, y, text, ha='center', va='center', fontsize=11, fontweight='bold')

    # Helper for flow arrows
    def draw_arrow(x, y_start, y_end):
        ax.annotate('', xy=(x, y_end), xytext=(x, y_start),
                    arrowprops=dict(facecolor='black', shrink=0.05, width=2, headwidth=10))

    # --- TIER 1: PERCEPTION (Classical) ---
    ax.add_patch(patches.Rectangle((0.5, 11.5), 9, 6.0, color='#FFF0F5', alpha=0.3))
    ax.text(5, 17.2, "TIER 1: PERCEPTION LAYER (Classical Processing)", ha='center', fontsize=14, fontweight='bold', color='purple')
    
    draw_box(5, 16, "INPUT\n$64 \\times 64$ RGB Traffic Sign Image", color='#ffffff')
    draw_arrow(5, 15.3, 14.7)
    draw_box(5, 14, "STN\nSpatial Transformer Network\n(Geometric Normalization)", color='#FFD1DC')
    draw_arrow(5, 13.3, 12.7)
    draw_box(5, 12, "PIFE\nGabor Filter Bank + CNN Refinement\n(Edge Feature Extraction)", color='#FFD1DC')

    # --- TIER 2: COGNITION (Quantum) ---
    ax.add_patch(patches.Rectangle((0.5, 6.5), 9, 4.5, color='#E6E6FA', alpha=0.3))
    ax.text(5, 10.7, "TIER 2: COGNITION LAYER (Quantum Transformation)", ha='center', fontsize=14, fontweight='bold', color='blue')
    
    draw_arrow(5, 11.3, 10.1)
    draw_box(5, 9.4, "8D LATENT BOTTLENECK (z)\nDense Angle Encoding", color='#C1CEFE')
    draw_arrow(5, 8.7, 8.1)
    draw_box(5, 7.4, "QUANTUM FEATURE TRANSFORMER (QFT)\n8-Qubit PQC (24 Trainable Parameters)", color='#C1CEFE')

    # --- TIER 3: EXECUTION (Robotic) ---
    ax.add_patch(patches.Rectangle((0.5, 0.5), 9, 5.5, color='#F0FFF0', alpha=0.3))
    ax.text(5, 5.7, "TIER 3: EXECUTION LAYER (Robotic Actuation)", ha='center', fontsize=14, fontweight='bold', color='green')
    
    draw_arrow(5, 6.7, 5.1)
    draw_box(5, 4.4, "LINEAR CLASSIFICATION HEAD\n(Mapping to 43 GTSRB Classes)", color='#C2F0C2')
    draw_arrow(5, 3.7, 3.1)
    draw_box(5, 2.4, "COMMAND LOGIC\nSerial/UART Protocol (Command Codes)", color='#C2F0C2')
    draw_arrow(5, 1.7, 1.1)
    draw_box(5, 0.4, "EMBEDDED ROBOT\nMicrocontroller -> PWM Adjust -> DC Motors", color='#C2F0C2')

    plt.savefig('HQCNN_Flowchart.png', dpi=300, bbox_inches='tight')
    print("Flowchart saved as HQCNN_Flowchart.png")

if __name__ == "__main__":
    draw_flowchart()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

def draw_padded_flowchart():
    # Large canvas to prevent any clipping at the bottom or sides
    fig, ax = plt.subplots(figsize=(12, 19))
    ax.set_xlim(0, 10)
    ax.set_ylim(-1, 21) 
    ax.axis('off')

    # Helper function for process boxes
    def draw_box(x, y, text, color='white', width=4.8, height=1.3):
        rect = patches.FancyBboxPatch((x - width/2, y - height/2), width, height, 
                                     boxstyle="round,pad=0.2", linewidth=2, 
                                     edgecolor='black', facecolor=color, zorder=3)
        ax.add_patch(rect)
        ax.text(x, y, text, ha='center', va='center', fontsize=11, fontweight='bold', zorder=4)

    # Helper for flow arrows
    def draw_arrow(x, y_start, y_end):
        ax.annotate('', xy=(x, y_end), xytext=(x, y_start),
                    arrowprops=dict(facecolor='black', shrink=0.05, width=2, headwidth=10), zorder=2)

    # --- TIER 1: PERCEPTION LAYER (Purple Background) ---
    # Background rectangle with extra padding (Slightly larger than the tier content)
    ax.add_patch(patches.Rectangle((0.2, 13.2), 9.6, 6.8, color='#FFF0F5', alpha=0.6, zorder=1, lw=1, ec='#DDA0DD'))
    ax.text(5, 19.4, "TIER 1: PERCEPTION LAYER (Classical Processing)", ha='center', fontsize=14, fontweight='bold', color='purple', zorder=2)
    
    draw_box(5, 18.2, "INPUT\n64 x 64 RGB Traffic Sign Image", color='#ffffff')
    draw_arrow(5, 17.5, 16.9)
    draw_box(5, 16.1, "STN\nSpatial Transformer Network\n(Geometric Normalization)", color='#FFD1DC')
    draw_arrow(5, 15.3, 14.7)
    draw_box(5, 14.2, "PIFE\nGabor Filter Bank + CNN Refinement\n(Edge Feature Extraction)", color='#FFD1DC')

    # Connection between Tier 1 and Tier 2
    draw_arrow(5, 13.2, 11.8)

    # --- TIER 2: COGNITION LAYER (Blue Background) ---
    # Background rectangle with extra padding
    ax.add_patch(patches.Rectangle((0.2, 7.8), 9.6, 5.0, color='#E6E6FA', alpha=0.6, zorder=1, lw=1, ec='#B0C4DE'))
    ax.text(5, 12.2, "TIER 2: COGNITION LAYER (Quantum Transformation)", ha='center', fontsize=14, fontweight='bold', color='blue', zorder=2)
    
    draw_box(5, 10.8, "8D LATENT BOTTLENECK (z)\nDense Angle Encoding", color='#C1CEFE')
    draw_arrow(5, 10.1, 9.5)
    draw_box(5, 8.8, "QUANTUM FEATURE TRANSFORMER (QFT)\n8-Qubit PQC (24 Trainable Parameters)", color='#C1CEFE')

    # Connection between Tier 2 and Tier 3
    draw_arrow(5, 7.8, 6.6)

    # --- TIER 3: EXECUTION LAYER (Green Background) ---
    # Background rectangle with extra padding
    ax.add_patch(patches.Rectangle((0.2, 0.0), 9.6, 7.2, color='#F0FFF0', alpha=0.6, zorder=1, lw=1, ec='#90EE90'))
    ax.text(5, 6.6, "TIER 3: EXECUTION LAYER (Robotic Actuation)", ha='center', fontsize=14, fontweight='bold', color='green', zorder=2)
    
    draw_box(5, 5.4, "LINEAR CLASSIFICATION HEAD\n(Mapping to 43 GTSRB Classes)", color='#C2F0C2')
    draw_arrow(5, 4.7, 4.1)
    draw_box(5, 3.4, "COMMAND LOGIC\nSerial/UART Protocol (Command Codes)", color='#C2F0C2')
    draw_arrow(5, 2.7, 2.1)
    draw_box(5, 1.4, "EMBEDDED ROBOT\nMicrocontroller -> PWM Adjust -> DC Motors", color='#C2F0C2')

    plt.tight_layout()
    plt.savefig('HQCNN_Final_Flowchart.png', dpi=300, bbox_inches='tight')
    plt.show()

draw_padded_flowchart()
```
